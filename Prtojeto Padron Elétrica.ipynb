```python
#Criar porjeto eletrico com base em orçamento para clientes diversos.
#Pedir dados dos clientes para coleta de dados e gerir orçamento personalizado.


print("=" * 60)
print("     SISTEMA DE ORÇAMENTO - Padron Elétrica")
print("=" * 60)

# Coletando dados do cliente e instalação
print("\n--- DADOS DO CLIENTE ---")
nome_cliente = input("Nome do cliente: ")
tipo_imovel = input("Tipo de imóvel (residencial/comercial/industrial): ").lower()

print("\n--- DADOS DA INSTALAÇÃO ---")
metros_quadrados = float(input("Área do imóvel (m²): "))
quantidade_comodos = int(input("Quantidade de cômodos: "))
projeto_eletrico = input("Possui projeto elétrico? (s/n): ").lower()

print("\n--- NECESSIDADES ESPECÍFICAS ---")
ar_condicionado = input("Vai instalar ar condicionado? (s/n): ").lower()
quantidade_ar = int(input("Quantos aparelhos de ar condicionado? ") if ar_condicionado == "s" else "0")
chuveiro_eletrico = input("Vai usar chuveiro elétrico? (s/n): ").lower()
automacao = input("Deseja automação residencial? (s/n): ").lower()

print("\n--- SEGURANÇA ---")
imovel_antigo = input("Imóvel é antigo (mais de 30 anos)? (s/n): ").lower()
fiacao_existente = input("Já possui reforma elétrica com menos de 5 anos? (s/n): ").lower()

print("\n" + "=" * 60)
print("CALCULANDO ORÇAMENTO E ANÁLISE TÉCNICA...")
print("=" * 60)

# Cálculo do custo base
custo_base = 0

# Custo por tipo de imóvel
if tipo_imovel == "residencial":
    custo_base = metros_quadrados * 85
    print(f"\n✓ Imóvel residencial: R$ {custo_base:.2f} (base)")
elif tipo_imovel == "comercial":
    custo_base = metros_quadrados * 120
    print(f"\n✓ Imóvel comercial: R$ {custo_base:.2f} (base - normas mais rigorosas)")
elif tipo_imovel == "industrial":
    custo_base = metros_quadrados * 200
    print(f"\n✓ Imóvel industrial: R$ {custo_base:.2f} (base - alta complexidade)")
else:
    print(f"\n❌ Tipo de imóvel inválido! Consulte um de nossos consultores.")
    exit()

# Acréscimos baseados nas necessidades
custo_adicional = 0

# Ar condicionado
if ar_condicionado == "s" and quantidade_ar > 0:
    custo_ar = quantidade_ar * 450
    custo_adicional += custo_ar
    print(f"✓ Instalação de {quantidade_ar} ar(es) condicionado(s): + R$ {custo_ar:.2f}")

    # Verificação de capacidade
    if quantidade_ar > quantidade_comodos:
        print(f"  ⚠️ ALERTA: {quantidade_ar} aparelhos para {quantidade_comodos} cômodos. Revisar necessidade!")
    elif quantidade_ar > 3 and tipo_imovel == "residencial":
        print(f"  ℹ️ DICA: Para {quantidade_ar} aparelhos, considere um sistema de ar centralizado")
else:
    print("✗ Sem ar condicionado")

# Chuveiro elétrico
if chuveiro_eletrico == "s":
    custo_chuveiro = 350
    custo_adicional += custo_chuveiro
    print(f"✓ Instalação de chuveiro elétrico: + R$ {custo_chuveiro:.2f}")

    # Verificação para múltiplos chuveiros
    if quantidade_comodos >= 2:
        print(f"  ℹ️ Para imóveis com mais de 1 banheiro, considere disjuntores individuais")
else:
    print("✗ Sem chuveiro elétrico")

# Automação residencial
if automacao == "s":
    custo_automacao = metros_quadrados * 35
    custo_adicional += custo_automacao
    print(f"✓ Automação residencial: + R$ {custo_automacao:.2f}")

    if tipo_imovel == "industrial":
        print(f"  🏭 Automação industrial requer engenheiro especializado")
    elif tipo_imovel == "comercial":
        print(f"  💼 Automação comercial inclui sistema de gerenciamento de energia")
else:
    print("✗ Sem automação")

# Projeto elétrico
if projeto_eletrico == "s":
    desconto_projeto = 300
    print(f"✓ Cliente possui projeto elétrico: - R$ {desconto_projeto:.2f} de desconto")
    custo_adicional -= desconto_projeto
else:
    custo_projeto = 500
    custo_adicional += custo_projeto
    print(f"✗ Sem projeto elétrico: + R$ {custo_projeto:.2f} (necessário elaborar projeto)")

# Verificação de segurança para imóveis antigos
custo_seguranca = 0
if imovel_antigo == "s":
    print("\n--- ANÁLISE DE SEGURANÇA PARA IMÓVEL ANTIGO ---")

    if fiacao_existente == "s":
        print("⚠️ ATENÇÃO: Imóvel antigo com fiação existente!")
        print("  → Recomendamos substituição completa da fiação")
        custo_seguranca += 1500
        print(f"  → Substituição de fiação: + R$ {custo_seguranca:.2f}")

        # Verificação adicional
        if tipo_imovel == "industrial":
            custo_seguranca += 2500
            print(f"  → Adequação às normas industriais: + R$ 2500.00")
    else:
        print("✓ Imóvel antigo sem fiação - ideal para instalação nova")
        custo_seguranca += 300
        print(f"  → Reforço estrutural para conduítes: + R$ {custo_seguranca:.2f}")

    # Alerta de segurança
    print("  🔴 ALERTA: Imóveis antigos geralmente têm carga elétrica insuficiente")
    print("  → Recomendamos upgrade do padrão de entrada para 3 fases")
    custo_seguranca += 800
    print(f"  → Upgrade padrão de entrada: + R$ 800.00")
else:
    if fiacao_existente == "n":
        print("✓ Imóvel novo sem fiação - instalação padrão")

# Custo total
custo_total = custo_base + custo_adicional + custo_seguranca

print("\n" + "=" * 60)
print("RESUMO DO ORÇAMENTO")
print("=" * 60)
print(f"Cliente: {nome_cliente}")
print(f"Tipo de imóvel: {tipo_imovel.upper()}")
print(f"Área: {metros_quadrados} m² | Cômodos: {quantidade_comodos}")
print("-" * 60)
print(f"Custo base: R$ {custo_base:.2f}")
print(f"Custos adicionais: R$ {custo_adicional:.2f}")
print(f"Segurança/adequação: R$ {custo_seguranca:.2f}")
print("-" * 60)
print(f"TOTAL DO ORÇAMENTO: R$ {custo_total:.2f}")
print("=" * 60)

# Recomendações finais baseadas em if/else
print("\n--- RECOMENDAÇÕES TÉCNICAS ---")

# Verifica necessidade de engenheiro
if tipo_imovel == "industrial":
    print("📌 Necessário engenheiro eletricista responsável (CREA)")
elif tipo_imovel == "comercial" and metros_quadrados > 300:
    print("📌 Recomendamos supervisão de engenheiro eletricista")
elif custo_total > 15000:
    print("📌 Projeto de alto valor - contrate um engenheiro para evitar retrabalho")
else:
    print("📌 Instalação pode ser realizada por equipe técnica qualificada")

# Verifica prazo de execução
if tipo_imovel == "residencial":
    if metros_quadrados < 100:
        prazo = 3
        print(f"📅 Prazo estimado: {prazo} dias úteis (instalação simples)")
    elif metros_quadrados < 250:
        prazo = 7
        print(f"📅 Prazo estimado: {prazo} dias úteis (instalação média)")
    else:
        prazo = 12
        print(f"📅 Prazo estimado: {prazo} dias úteis (instalação complexa)")
elif tipo_imovel == "comercial":
    prazo = 15
    print(f"📅 Prazo estimado: {prazo} dias úteis (comercial exige mais atenção)")
else:  # industrial
    prazo = 30
    print(f"📅 Prazo estimado: {prazo} dias úteis (industrial - alta complexidade)")

# Verifica necessidade de projeto especial
if automacao == "s" and ar_condicionado == "s":
    print("🤖 RECOMENDAÇÃO ESPECIAL: Sistema integrado de automação + ar condicionado")
    print("   → Isso pode gerar economia de energia de até 30%")

    if tipo_imovel == "comercial":
        print("   → Considere sistema de gestão de energia (SGE)")

# Aprovação ou sugestão de financiamento
print("\n" + "=" * 60)
print("CONCLUSÃO:")
print("=" * 60)

if custo_total > 20000:
    if tipo_imovel == "residencial":
        print("💰 Valor elevado para residência. Oferecemos:")
        print("  → Parcelamento em até 12x sem juros")
        print("  → Desconto de 5% para pagamento à vista")
    else:
        print("💰 Orçamento compatível com o porte do empreendimento")
        print("  → Condições especiais para empresas (consulte comercial)")
elif custo_total > 10000:
    print("💰 Orçamento médio. Prazo de execução confiável.")
    print("  → Parcelamento em até 6x sem juros")
else:
    print("💰 Orçamento acessível. Ideal para execução imediata!")
    print("  → Pagamento à vista com 10% de desconto!")

# Verificação final de segurança
print("\n--- CHECKLIST DE SEGURANÇA ---")

if imovel_antigo == "s" and fiacao_existente == "s":
    print("❌ CRÍTICO: Necessário laudo de segurança da fiação existente")
    print("   → Contrate urgente uma vistoria técnica")
elif imovel_antigo == "s" and fiacao_existente == "n":
    print("⚠️ ATENÇÃO: Será necessário projeto de adequação de carga")
elif chuveiro_eletrico == "s" and tipo_imovel == "residencial" and quantidade_comodos == 1:
    print("✓ Padrão: Chuveiro em kitnet - dimensionamento padrão")
else:
    print("✓ Instalação dentro dos padrões de segurança")

print("\n" + "=" * 60)
print("ORÇAMENTO VÁLIDO POR 30 DIAS")
print(f"Padron Elétrica - {nome_cliente}, aguardamos sua decisão!")
print("=" * 60)
```
