```python
print("=" * 40)
print("     SISTEMA DE DESCONTOS")
print("=" * 40)

# Entrada de dados
valor_compra = float(input("Valor total da compra: R$ "))
cliente_vip = input("Cliente VIP? (s/n): ").lower()
forma_pagamento = input("Forma de pagamento (dinheiro/cartão/pix): ").lower()
dia_promocao = input("Hoje é dia de promoção? (s/n): ").lower()

print("\n" + "-" * 40)

# Cálculo do desconto
desconto = 0

# Regra 1: Cliente VIP ganha 5% de desconto
if cliente_vip == "s":
    desconto += 5
    print("✓ Cliente VIP: +5% de desconto")
else:
    print("✗ Cliente regular")

# Regra 2: Desconto extra para formas de pagamento específicas
if forma_pagamento == "dinheiro":
    desconto += 7
    print("✓ Pagamento em dinheiro: +7% de desconto")
elif forma_pagamento == "pix":
    desconto += 5
    print("✓ Pagamento via PIX: +5% de desconto")
elif forma_pagamento == "cartao":
    print("✓ Pagamento com cartão: sem desconto extra")
else:
    print("✗ Forma de pagamento inválida")

# Regra 3: Desconto por dia de promoção
if dia_promocao == "s":
    desconto += 10
    print("✓ Dia de promoção: +10% de desconto")
elif dia_promocao == "n":
    print("✗ Sem promoção no dia")

# Regra 4: Desconto especial para compras acima de R$ 500
if valor_compra > 500:
    desconto_extra = 5
    desconto += desconto_extra
    print(f"✓ Compra acima de R$500: +{desconto_extra}% de desconto")

# Limitar desconto máximo a 30%
if desconto > 30:
    print(f"\n⚠️ Desconto original de {desconto}% limitado a 30%")
    desconto = 30

# Cálculo final
valor_desconto = valor_compra * (desconto / 100)
valor_final = valor_compra - valor_desconto

print("\n" + "=" * 40)
print("RESUMO DA COMPRA")
print("=" * 40)
print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Desconto total: {desconto}%")
print(f"Valor do desconto: R$ {valor_desconto:.2f}")

# Condição final com if/else
if valor_final > 0:
    print(f"Valor a pagar: R$ {valor_final:.2f}")

    # Verifica se vale a pena comprar agora
    if desconto >= 20:
        print("\n🎉 Excelente compra! Você conseguiu um ótimo desconto!")
    elif desconto >= 10:
        print("\n👍 Bom desconto! Aproveite!")
    elif desconto > 0:
        print("\n👌 Desconto pequeno, mas é melhor que nada!")
    else:
        print("\n😕 Sem desconto dessa vez... Tente outro dia!")
else:
    print("\n⚠️ ERRO: O valor final não pode ser negativo!")

print("=" * 40)
```
